```python
# CASE-BASED LEARNING

# Previous cases
cases = [
    {"hours": 2, "mark": 45},
    {"hours": 4, "mark": 60},
    {"hours": 6, "mark": 75},
    {"hours": 8, "mark": 88}
]

# New student
new_hours = 6.8

# Initially no best case
best_case = None

# Start with infinity
smallest_distance = float("inf")

# Compare new student
# with every previous case

for case in cases:

    # Calculate distance
    distance = abs(
        case["hours"] - new_hours
    )

    print(
        "Previous Case:",
        case["hours"],
        "hours | Distance:",
        distance
    )

    # Check whether this is
    # the closest case
    if distance < smallest_distance:

        smallest_distance = distance

        best_case = case

# Display result
print(
    "\nNew Student:",
    new_hours,
    "hours"
)

print(
    "Most Similar Previous Case:",
    best_case["hours"],
    "hours"
)

print(
    "Previous Student Mark:",
    best_case["mark"]
)

print(
    "Suggested Predicted Mark:",
    best_case["mark"]
)
```
